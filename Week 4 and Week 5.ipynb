```python
from itertools import combinations

def calculate_cost(state):
    cost = 0
    for i in range(4):
        for j in range(i + 1, 4):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost

def generate_neighbours(state):
    neighbours = []
    for i, j in combinations(range(4), 2):
        new_state = state.copy()
        new_state[i], new_state[j] = new_state[j], new_state[i]
        neighbours.append(new_state)
    return neighbours

def hill_climbing():
    current = list(map(int, input("Enter initial state (4 numbers): ").split()))

    if len(current) != 4 or sorted(current) != [0, 1, 2, 3]:
        print("Invalid input! Enter a permutation of 0, 1, 2, 3.")
        return

    iteration = 1

    while True:
        current_cost = calculate_cost(current)

        print("\nIteration:", iteration)
        print("Current State:", current)
        print("Cost:", current_cost)

        if current_cost == 0:
            print("Goal State Reached!")
            break

        neighbours = generate_neighbours(current)

        best_state = min(neighbours, key=calculate_cost)
        best_cost = calculate_cost(best_state)

        if best_cost < current_cost:
            current = best_state
        else:
            print("No better neighbour found!")
            print("Algorithm Terminated.")
            break

        iteration += 1

    print("\nFinal State:", current)
    print("Final Cost:", calculate_cost(current))


hill_climbing()
```

```python
import random
import math

def calculate_cost(state):
    cost = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost

def generate_neighbour(state):
    neighbour = state.copy()
    col = random.randint(0, 7)
    new_row = random.randint(0, 7)
    while new_row == neighbour[col]:
        new_row = random.randint(0, 7)
    neighbour[col] = new_row
    return neighbour

def simulated_annealing():

    current = [random.randint(0, 7) for i in range(8)]
    T = 100
    cooling_rate = 0.995
    iteration = 1
    print("Initial State:", current)
    print("Initial Cost:", calculate_cost(current))
    while T > 0.1 and iteration <= 2:
        current_cost = calculate_cost(current)

        if current_cost == 0:
            break

        next_state = generate_neighbour(current)
        next_cost = calculate_cost(next_state)

        delta_E = current_cost - next_cost

        if delta_E > 0:
            probability = 1
            current = next_state
            accepted = True
        else:
            probability = math.exp(delta_E / T)
            if random.random() < probability:
                current = next_state
                accepted = True
            else:
                accepted = False

        print("\nIteration:", iteration)
        print("Current Cost:", current_cost)
        print("Neighbour State:", next_state)
        print("Neighbour Cost:", next_cost)
        print("Delta E:", delta_E)
        print("Temperature:", round(T, 4))
        print("Acceptance Probability:", round(probability, 4))
        print("Accepted:", accepted)
        print("New Current State:", current)

        T = T * cooling_rate
        iteration += 1
    print("\n========== FINAL RESULT ==========")
    final_cost = calculate_cost(current)
    print("Final State:", current)
    print("Final Cost:", final_cost)
    if final_cost == 0:
        print("Goal State Reached!")
    else:
        print("Solution not found within the temperature limit.")
simulated_annealing()
```
