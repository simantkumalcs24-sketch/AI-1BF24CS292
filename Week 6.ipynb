```python

from itertools import product

symbols = input("Enter symbols: ").split()

print("\nUse: and, or, not")
kb = input("Enter Knowledge Base: ")

query = input("Enter Query: ")

print("\nTruth Table:")
print(*symbols, "KB", "Query", sep="\t")

entailed = True

for values in product([True, False], repeat=len(symbols)):
    model = dict(zip(symbols, values))

    KB = eval(kb, {"__builtins__": {}}, model)
    Q = eval(query, {"__builtins__": {}}, model)

    print(*values, KB, Q, sep="\t")

    if KB and not Q:
        entailed = False

print("\nResult:")
if entailed:
    print("Query is ENTAILED by KB")
else:
    print("Query is NOT ENTAILED by KB")
```
